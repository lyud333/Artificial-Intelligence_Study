```python
f=open("name.txt", encoding="utf-8")
data=f.read()
print(data)
f.close()

# r 읽기(기본값, 생략) : 파일이 반드시 존재해야함. 파일이 없으면 오류
# w 쓰기 : 파일이 이미 존재하면 기존 내용을 지우고 새로 작성, 파일이 없으면 새 파일 생성
# a 추가 : 파일이 존재하면 기존 내용의 끝에 추가, 파일이 없으면 새로 생성
```

```python
f=open("name.txt", 'w')
while True:
    team=input("팀: ")
    if not team:
        break
    f.write(team + "\n")
f.close()
#write() : 데이터를 파일에 문자로 기록, 자동 줄바꿈이 되지 않음
```

```python
import pandas as pd

data = {
    '이름' : ['유비', '관우', '장비'],
    '1차지필' : [92, 89, 71],
    '2차지필' : [88, 90, 84],
    '수행1'  : [93, 85, 67],
    '수행2'  : [95, 78, 70],
    '등급'   : ['A', 'B', 'C']
}
df = pd.DataFrame(data)
df
df.to_excel("my_grade.xlsx", index=False)
```

```python

df=pd.read_excel("my_grade.xlsx")
df.head()

# pd.read_excel() : 엑셀 파일을 읽어 DataFrame객체로 반환
# head() 또는 tail() : DataFrame의 상위 또는 하위 5개 행을 반환

add_df=pd.DataFrame([{
    '이름' : '조자룡',
    '1차지필' : 63,
    '2차지필' : 75,
    '수행1' : 84,
    '수행2' : 79,
    '등급' : 'C'
}])
modi_file=pd.concat([df, add_df], ignore_index=True)
modi_file.to_excel("new_grade.xlsx", index=False)
modi_file

# concat() : 두 개의 데이터 프레임을 합침
# to_excel() : DataFrame객체를 엑셀 파일로 저장
# index=False : 인덱스는 저장하지 않음
```

```python
import pandas as pd
df=pd.read_excel("club.xlsx")
df
add_df=pd.DataFrame([{
    '학번': 10512,
    '이름': '김하늘',
    '학년': 1,
    '동아리': 'AI연구부',
    '역할': '부원'
}])
new_df=pd.concat([df, add_df], ignore_index=True)
# index=True : 인덱스를 새로 생성, index=False : 기존 인덱스를 유지
new_df.to_excel("new_club.xlsx", index=False)
new_df
```
