```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('student.csv')
df.head()
```

```python
plt.hist(df['StudyHours'], bins=10, color='blue', alpha=0.7)
plt.xlabel('Study Hours')
plt.ylabel('Assignment Scores')
plt.title('StudyHours vs Assignment Scores')
plt.show()
```

```python
# 공부시간과 과제점수의 관계 알아보기 : 산점도

plt.scatter(df['StudyHours'], df['AssignmentScore'], color='blue', alpha=0.7)
plt.xlabel('Study Hours')
plt.ylabel('Assignment Scores')
plt.title('StudyHours vs Assignment Scores')
plt.show()
```

```python
result_mean=df.groupby('Result')['StudyHours'].mean()
print(result_mean)
```

```python
plt.bar(result_mean.index, result_mean.values, color='blue', alpha=0.7)
plt.xticks([0, 1])
plt.ylabel('Average Study Hours')   
plt.title('Average Study Hours by Result')
plt.show()
```

```python
# 개별화 1번
#출석률과 과제 점수의 관계 알아보기 : 산점도
plt.scatter(df['Attendance'], df['AssignmentScore'], color='blue', alpha=0.7)
plt.xlabel('Attendance')
plt.ylabel('Assignment Scores')
plt.title('Attendance vs Assignment Scores')
plt.show()
```

```python
# 개별화 3번
#수면시간과 공부시간의 관계 알아보기 : 히스토그램

plt.hist(df['SleepHours'], bins=10, color='blue', alpha=0.7)
plt.xlabel('Sleep Hours')
plt.ylabel('Study Hours')
plt.title('SleepHours vs Study Hours')
plt.show()
```
