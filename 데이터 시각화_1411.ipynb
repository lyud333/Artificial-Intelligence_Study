# 초미세먼지 데이터 시각화

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel('dust_pm25.xlsx', index_col='area')
print(data)


plt.figure(figsize=(9,3))

for year in range(2019,2022):
    y_data = data[year]
    plt.plot(y_data, marker='s', label=year)

plt.xlabel('area')
plt.ylabel('micrometer')
plt.xticks(rotation=45)
plt.title('2019~2021 Dust Line Graph')
plt.legend()
plt.show()
```

```python
# 초미세먼지 데이터의 2017년 자료의 빈도수
data2017=data[2017]
plt.hist(data2017, bins=9, label='pm2.5')
plt.xlabel('pm2.5')
plt.ylabel('frequency')
plt.title('2017 Dust Histogram')
plt.legend()
plt.grid()
plt.show()
```
